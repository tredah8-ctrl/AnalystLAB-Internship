**How do host characteristics and performance indicators vary across cities?**

```python
 # Import pandas
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. HOST RESPONSE TIME BY CITY
response_time_summary = (
    listing[listing["city"] != "null"]
    .groupby(["city", "host_response_time"])
    .size()
    .reset_index(name="listing_count")
)

print("HOST RESPONSE TIME BY CITY")
display(response_time_summary)

# 2. HOST RESPONSE RATE BY CITY
# Convert response rate to numeric after removing the percentage sign
listing["host_response_rate"] = (
    listing["host_response_rate"]
    .astype(str)
    .str.replace("%", "", regex=False)
)

listing["host_response_rate"] = pd.to_numeric(
    listing["host_response_rate"], errors="coerce"
)

response_rate_summary = (
    listing[listing["city"] != "null"]
    .groupby("city")
    .agg(
        average_response_rate=("host_response_rate", "mean"),
        listing_count=("listing_id", "count")
    )
    .sort_values("average_response_rate", ascending=False)
)

print("AVERAGE HOST RESPONSE RATE BY CITY")
display(response_rate_summary)

# 3. SUPERHOST STATUS BY CITY
superhost_summary = (
    listing[listing["city"] != "null"]
    .groupby(["city", "host_is_superhost"])
    .size()
    .reset_index(name="listing_count")
)

print("SUPERHOST STATUS BY CITY")
display(superhost_summary)

# 4. HOST EXPERIENCE BY CITY
# Convert host_since into a date
listing["host_since"] = pd.to_datetime(
    listing["host_since"], errors="coerce"
)

host_experience_summary = (
    listing[listing["city"] != "null"]
    .groupby("city")
    .agg(
        average_host_start_date=("host_since", "mean"),
        unique_hosts=("host_id", "nunique")
    )
)

print("HOST EXPERIENCE INFORMATION BY CITY")
display(host_experience_summary)
```

```python

# Machine Learning tools
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Make a copy so we do not change the original dataset
model_data = listing.copy()

# 2. Remove listings without a valid price
model_data = model_data.dropna(subset=["price"])

# 3. Remove rows where the city is recorded as "null"
model_data = model_data[model_data["city"] != "null"]

# 4. Select useful numerical characteristics
numeric_candidates = [
    "accommodates",
    "minimum_nights",
    "maximum_nights",
    "host_total_listings_count",
    "review_scores_rating",
    "host_response_rate",
    "host_acceptance_rate"
]

# Keep only numerical columns that exist in the dataset
numeric_features = [
    col for col in numeric_candidates
    if col in model_data.columns
]

# Convert numerical features into numeric values
for col in numeric_features:
    model_data[col] = pd.to_numeric(
        model_data[col].astype(str).str.replace("%", "", regex=False),
        errors="coerce"
    )

# 5. Select useful categorical characteristics
categorical_candidates = [
    "city",
    "room_type",
    "property_type",
    "host_is_superhost",
    "host_response_time",
    "instant_bookable"
]

# Keep only categorical columns that exist
categorical_features = [
    col for col in categorical_candidates
    if col in model_data.columns
]

# 6. Combine all selected features
features = numeric_features + categorical_features

# X contains the information used to make predictions
X = model_data[features]

# y contains the price we want the model to predict
y = model_data["price"]

print("FEATURES USED:")
print(features)

print("NUMBER OF LISTINGS:", len(model_data))

# 7. Divide the data into training and testing sets
# 80% is used to train the model; 20% is used to test it
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# 8. Prepare numerical and categorical data
# Missing numerical values are replaced with the median
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

# Missing categorical values are replaced with the most common value
# OneHotEncoder converts categories into numerical columns
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Apply the correct preparation to each feature type
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

# 9. Create the Random Forest Regression model
model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        min_samples_leaf=2
    ))
])

# 10. Train the model
model.fit(X_train, y_train)

# 11. Generate predictions using the test data
predictions = model.predict(X_test)

# 12. Evaluate model performance
mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print("RANDOM FOREST REGRESSION RESULTS")
print("--------------------------------")
print("Mean Absolute Error (MAE):", mae)
print("Root Mean Squared Error (RMSE):", rmse)
print("R-squared (R2):", r2)

# 13. Compare actual prices with predicted prices
comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": predictions
})

print("ACTUAL VS PREDICTED PRICES")
display(comparison.head(10))
```

A Kruskal-Wallis test was conducted to determine whether advertised prices differed across room types in Cape Town. The results indicated a statistically significant difference in price distributions, H(3) = 48.05, p < 0.001, with an effect size of approximately 0.089. Therefore, the null hypothesis was rejected.
```python
 # Prepare the price data for each room type
entire_place = cape_town[cape_town['room_type'] == 'Entire place']['price'].dropna()
hotel_room = cape_town[cape_town['room_type'] == 'Hotel room']['price'].dropna()
private_room = cape_town[cape_town['room_type'] == 'Private room']['price'].dropna()
shared_room = cape_town[cape_town['room_type'] == 'Shared room']['price'].dropna()

# Perform the Kruskal-Wallis statistical test
statistic, p_value = kruskal(
    entire_place,
    hotel_room,
    private_room,
    shared_room
)

# Calculate the total number of observations and groups
n = len(entire_place) + len(hotel_room) + len(private_room) + len(shared_room)
k = 4

# Calculate the effect size (epsilon squared)
effect_size = (statistic - k + 1) / (n - k)

# Display the results
print("Kruskal-Wallis Test Results")
print("---------------------------")
print("Test statistic:", statistic)
print("P-value:", p_value)
print("Effect size:", effect_size)

# Interpret the statistical result using a 5% significance level
if p_value < 0.05:
    print("Decision: Reject the null hypothesis.")
    print("There is a statistically significant difference in prices between room types.")
else:
    print("Decision: Fail to reject the null hypothesis.")
    print("There is insufficient evidence of a statistically significant difference in prices.")
```

```python
 # Perform the Mann-Whitney U test
statistic, p_value = mannwhitneyu(
    entire_ratings,
    private_ratings,
    alternative='two-sided'
)

# Calculate the total number of observations
n1 = len(entire_ratings)
n2 = len(private_ratings)

# Calculate the Mann-Whitney effect size (rank-biserial correlation)
effect_size = 1 - (2 * statistic) / (n1 * n2)

# Display the statistical test results
print("Mann-Whitney U Test Results")
print("----------------------------")
print("U statistic:", statistic)
print("P-value:", p_value)
print("Effect size:", effect_size)

# Interpret the p-value at a 5% significance level
if p_value < 0.05:
    print("Decision: Reject the null hypothesis.")
    print("There is a statistically significant difference in ratings.")
else:
    print("Decision: Fail to reject the null hypothesis.")
    print("There is insufficient evidence of a significant difference.")
```

The three tests provide evidence that room type is associated with advertised price, accommodation capacity is positively associated with advertised price, and overall rating distributions differ between entire-place and private-room listings in Cape Town.

These findings may help accommodation providers understand how property characteristics relate to advertised prices and guest ratings. However, statistical significance does not necessarily mean that the differences are commercially large. The analysis is observational and cannot establish causation.

Important limitations across the three hypotheses:
1. The price comparisons use Cape Town only to avoid currency differences.
2. Small room-type groups may produce less reliable estimates.
3. The analysis does not control for all possible influences, such as neighbourhood, amenities or property size.
4. Advertised prices are not actual booking prices or revenue.
5. The reviews-to-listings ID mismatch remains unresolved, so we have not used that relationship to draw conclusions about review activity by listing category.
