# Task 1: Define the Research Question

## Research Question

How do encounter costs vary across different encounter classes?

## Relevance of the Research Question

Healthcare encounters involve different types of services, including emergency visits, inpatient admissions, ambulatory visits and wellness visits. These encounter types may have different recorded costs depending on the nature and complexity of the services provided.

Investigating cost differences across encounter classes can help identify which types of encounters have higher recorded costs and provide useful insights into healthcare expenditure.

## Variables Required

The following variables from the encounters dataset will be used:

- **EncounterClass:** Identifies the type of healthcare encounter.
- **Base_Encounter_Cost:** Records the base cost of the encounter, excluding additional line-item costs.
- **Total_Claim_Cost:** Records the total encounter cost, including additional services.
- **Payer_Coverage:** Records the amount covered by the healthcare payer and provides supporting information for the analysis.

The primary analysis will compare total claim costs across encounter classes, with base encounter costs used as an additional measure.
